```python
# Assignment - Data Visualization + Challenge / Extension Tasks

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# -------------------------------------------------
# 1. Allow user to provide the CSV file name as input
# -------------------------------------------------
file_name = input("Enter the CSV file name (e.g., sample_data_assignment_07.csv): ")
print("The CSV file name is: "+file_name)

# Load the CSV dataset
df = pd.read_csv(file_name)

# Display the dataset
print("\nDataset:")
print(df)
print("\nColumns available:", df.columns.tolist())

# -------------------------------------------------
# 2. Calculate and display average marks
# -------------------------------------------------
print("\n========== Average Values ==========")
print("Average Maths Marks     :", round(df["Maths"].mean(), 2))
print("Average Python Marks    :", round(df["Python"].mean(), 2))
print("Average Attendance      :", round(df["Attendance"].mean(), 2))

# -------------------------------------------------
# 3. Line Plot (Python Marks)
# -------------------------------------------------
plt.figure(figsize=(8, 5))
plt.plot(df["Student"], df["Python"], marker="o", color="blue", label="Python Marks")
plt.title("Python Marks of Students", fontsize=14)
plt.xlabel("Student")
plt.ylabel("Python Marks")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# -------------------------------------------------
# 4. Bar Plot (Maths Marks)
# -------------------------------------------------
plt.figure(figsize=(8, 5))
plt.bar(df["Student"], df["Maths"], color="orange", label="Maths Marks")
plt.title("Maths Marks of Students", fontsize=14)
plt.xlabel("Student")
plt.ylabel("Maths Marks")
plt.legend()
plt.tight_layout()
plt.show()

# -------------------------------------------------
# 5. Scatter Plot (Maths vs Python)
# -------------------------------------------------
plt.figure(figsize=(7, 5))
plt.scatter(df["Maths"], df["Python"], s=80, color="green", label="Students")
plt.title("Maths vs Python Marks", fontsize=14)
plt.xlabel("Maths Marks")
plt.ylabel("Python Marks")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# -------------------------------------------------
# 6. Histogram of Python Marks
# -------------------------------------------------
plt.figure(figsize=(7, 5))
plt.hist(df["Python"], bins=8, color="skyblue", edgecolor="black", label="Python Marks")
plt.title("Histogram of Python Marks", fontsize=14)
plt.xlabel("Python Marks")
plt.ylabel("Number of Students")
plt.legend()
plt.grid(axis="y", alpha=0.5)
plt.tight_layout()
plt.show()

# -------------------------------------------------
# 7. Box Plot for Maths and Python Marks
# -------------------------------------------------
plt.figure(figsize=(7, 5))
sns.boxplot(data=df[["Maths", "Python"]], palette="Set2")
plt.title("Box Plot of Maths and Python Marks", fontsize=14)
plt.ylabel("Marks")
plt.tight_layout()
plt.show()

# -------------------------------------------------
# 8. Pie Chart – Number of students from each city
# -------------------------------------------------
if "City" in df.columns:
    city_counts = df["City"].value_counts()
    
    plt.figure(figsize=(7, 7))
    plt.pie(city_counts, labels=city_counts.index, autopct="%1.1f%%",
            startangle=90, colors=sns.color_palette("pastel"),
            wedgeprops={"edgecolor": "white"})
    plt.title("Number of Students from Each City", fontsize=14)
    plt.legend(title="City", loc="best")
    plt.tight_layout()
    plt.show()
else:
    print("\nNote: 'City' column not found in the dataset. Skipping Pie Chart.")

# -------------------------------------------------
# 9. Correlation Matrix and Heatmap
# -------------------------------------------------
correlation = df[["Maths", "Python", "Attendance"]].corr()
print("\nCorrelation Matrix:")
print(correlation)

plt.figure(figsize=(6.5, 5))
sns.heatmap(correlation, annot=True, cmap="Blues", vmin=-1, vmax=1, linewidths=0.5)
plt.title("Correlation Heatmap", fontsize=14)
plt.tight_layout()
plt.show()

print("\n========== All visualizations completed successfully ==========")
```
