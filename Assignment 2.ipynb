```python
#Taking input of three numbers
print("Enter first number:")
a=int(input())
print("Enter second number:")
b=int(input())
print("Enter third number:")
c=int(input())

#Logic for finding the largest number among the three
if a>=b and a>=c:
    largest=a
elif b>=a and b>=c:
    largest=b
else:
    largest=c
print("The largest number among the three is:",largest)
```
