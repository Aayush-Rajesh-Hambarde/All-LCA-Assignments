```python
#Taking input of three numbers
print("Enter first number:")
a=int(input())
print("\nThe first number is:",a)
print("\nEnter second number:")
b=int(input())
print("\nThe second number is:",b)
print("\nEnter third number:")
c=int(input())
print("\nThe third number is:",c)

#Logic for finding the largest number among the three
if a>=b and a>=c:
    largest=a
elif b>=a and b>=c:
    largest=b
else:
    largest=c
print("\nThe largest number among the three is:",largest)
```
