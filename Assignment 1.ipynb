```python
#Creating a Dictionary of students
stu_dict={1:"Rahul",2:"Priya",3:"Rakesh",4:"Rajesh",5:"Rohan"}
#Creating a Tuple of students
stu_tuple=("Amit","Raj","Amit","Rahil","Amit")
#Creating a List of students
stu_list=["Ritesh","Aarav","Harsh","Aditya","Advait"]

#Performing List Operations
print("List operations:")

print("\nOriginal List:")
print(stu_list)
stu_list.append("Ramesh")
print("\nList after appending Ramesh:")
print(stu_list)
stu_list.insert(2,"Zahir")
print("\nList after inserting Zahir at Index 2:")
print(stu_list)
stu_list.remove("Ritesh")
print("\nList after removing Ritesh:")
print(stu_list)
print("\nReturning the element from Index 4 of List using pop method:")
print(stu_list.pop(4))
print("\nList after removing element from Index 4 using pop method:")
print(stu_list)
stu_list.sort()
print("\nList after sorting it:")
print(stu_list)
print()

#Performing Tuple Operations
print("\nTuple operations:")

print("\nOriginal Tuple:")
print(stu_tuple)
print("\nNumber of times Amit appears in the Tuple is:",stu_tuple.count("Amit"))
print("\nThe Index of Raj in the Tuple is:",stu_tuple.index("Raj"))
print("\nThe number of names in the Tuple is:",len(stu_tuple))
print("\nThe sorted Tuple is:",sorted(stu_tuple))
print("\nThe name that comes first alphabetically in the Tuple is:",min(stu_tuple))
print("\nThe name that comes last alphabetically in the Tuple is:",max(stu_tuple))
print()

#Performing Dictionary Operations
print("\nDictionary operations:")

print("\nOriginal Dictionary:")
print(stu_dict)
print("\nAll the keys in the Dictionary:")
print(stu_dict.keys())
print("\nAll the values in the Dictionary:")
print(stu_dict.values())
print("\nAll the items in the Dictionary:")
print(stu_dict.items())
print("\nThe value associated with the key 2 in the Dictionary is:",stu_dict.get(2))
print("\nRemoving and returning the value associated with the key 3 in the Dictionary using pop is:",stu_dict.pop(3))
print("\nThe Dictionary after using pop is:")
print(stu_dict)
```
