```python
#To find if triangle is right angled or not
a=float(input("Enter the first side of the triangle:"))
print("\nThe first side of the triangle is:",a)
b=float(input("\nEnter the second side of the triangle:"))
print("\nThe second side of the triangle is:",b)
c=float(input("\nEnter the third side of the triangle:"))
print("\nThe third side of the triangle is:",c)

sides=sorted([a,b,c])
if a<=0 or b<=0 or c<=0:
    print("\nThe triangle does not exist")
elif sides[0]+sides[1]>sides[2]:
    if sides[0]**2+sides[1]**2==sides[2]**2:
        print("\nThe triangle is a right angled triangle")
    else:
        print("\nThe triangle is not a right angled triangle")
else:
    print("\nThe triangle is not a right angled triangle")
```
