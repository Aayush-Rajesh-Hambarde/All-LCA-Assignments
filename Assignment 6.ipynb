```python
# Assignment No. 6
# Data Preprocessing on .csv / .xls file using Pandas

import pandas as pd
import numpy as np

# -------------------------------------------------
# 1. Create a sample DataFrame (or you can read from CSV)
# -------------------------------------------------
data = {
    'Name': ['Amit', 'Priya', 'Rahul', 'Sneha', 'Amit', 'Vikas', 'Neha', 'Rahul', None],
    'Marks': [85, 90, np.nan, 78, 85, 92, np.nan, 88, 75],
    'City': ['Pune', 'Mumbai', 'Pune', 'Delhi', 'Pune', 'Mumbai', 'Delhi', 'Pune', 'Nagpur']
}

df = pd.DataFrame(data)

print("========== Original DataFrame ==========")
print(df)
print("\nShape of original data:", df.shape)
print("\nMissing values in each column:")
print(df.isnull().sum())

# -------------------------------------------------
# 2. Handle Missing Values
# -------------------------------------------------

# Method 1: Drop rows that have any missing values
df_dropped = df.dropna()
print("\n========== After dropna() ==========")
print(df_dropped)

# Method 2: Fill missing values
df_filled = df.copy()
df_filled['Marks'] = df_filled['Marks'].fillna(df_filled['Marks'].mean())   # Fill Marks with mean
df_filled['Name'] = df_filled['Name'].fillna('Unknown')                    # Fill Name with 'Unknown'

print("\n========== After fillna() ==========")
print(df_filled)

# -------------------------------------------------
# 3. Remove Duplicate Records
# -------------------------------------------------
print("\n========== Checking Duplicates ==========")
print("Duplicate rows:\n", df_filled.duplicated())

# Remove duplicate rows
df_no_duplicates = df_filled.drop_duplicates()
print("\n========== After removing duplicates ==========")
print(df_no_duplicates)

# -------------------------------------------------
# 4. Data Selection and Filtering using loc[]
# -------------------------------------------------

# Select specific rows and columns using loc
print("\n========== Using loc[] ==========")

# Select rows where Marks > 80 and only Name & Marks columns
high_scorers = df_no_duplicates.loc[df_no_duplicates['Marks'] > 80, ['Name', 'Marks']]
print("Students with Marks > 80:")
print(high_scorers)

# Select all students from Pune
pune_students = df_no_duplicates.loc[df_no_duplicates['City'] == 'Pune']
print("\nStudents from Pune:")
print(pune_students)

# -------------------------------------------------
# 5. Optional: Save the cleaned data to a CSV file
# -------------------------------------------------
df_no_duplicates.to_csv('cleaned_student_data.csv', index=False)
print("\nCleaned data saved to 'cleaned_student_data.csv'")

print("\n========== Final Cleaned DataFrame ==========")
print(df_no_duplicates)
```
