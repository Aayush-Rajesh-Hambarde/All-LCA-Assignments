```python
print("\nColumns:")
df.columns
```

```python
#Step 11: Display passenger names
print("\nPassenger Names:")
print(df["Name"].head())

#Step 12: Count Passengers by class
print("\nPassengers by Gender:")
print(df["Pclass"].value_counts())

#Step 13: Calculate Average Age
print("\nAverage Age:")
print(df["Age"].mean())

#Step 14: Find Highest fare
print("\nHighest Fare:")
print(df["Fare"].max())

#Step 15: Find Average Fare
print("\nAverage Fare:")
print(df["Fare"].mean())
```

```python
#Step 16: Filter passengers older than 60
print("\nPassengers older than 60:")
print(df.loc[df["Age"]>60,"Age"])
```

```python
#Step 17: Display number of passengers who survived
print("\nNumber of passengers who survived:")
print(df["Survived"].value_counts())
```

```python
#Step 18: Display passengers who survived
print("\nDisplay passengers who survived:")
print(df[df["Survived"]==1][["Name","Sex","Age"]].head())
```

```python
#Step 19: Calculate Survival Rate by Gender
print("\nSurvival rate by Gender")
print(df.groupby("Sex")["Survived"].mean())
```

```python
#Step 20: Sort passengers by Fare
print("\nTop 10 highest fares:")
print(df.sort_values("Fare",ascending=False)[["Name","Fare"]].head(10))
```
