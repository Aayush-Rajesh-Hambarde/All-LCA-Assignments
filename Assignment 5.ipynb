```python
import re

# Accept string from user
s = input("Enter a string: ")
print("The string inputed is: "+s)

# Regular expression pattern: only a-z, A-Z, 0-9
pattern = r'^[a-zA-Z0-9]+$'

# Check the string
if re.match(pattern, s):
    print("Valid: The string contains only letters and digits.")
else:
    print("Invalid: The string contains other characters.")
```

```python
# Program to check if a string contains only a-z, A-Z and 0-9

# Accept string from user
user_string = input("Enter a string: ")
print("The string inputed is: "+user_string)

# Check if the string contains only alphanumeric characters
if user_string.isalnum():
    print("The string contains only a-z, A-Z and 0-9 characters.")
else:
    print("The string contains characters other than a-z, A-Z and 0-9.")
```
